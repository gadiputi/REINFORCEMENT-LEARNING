```python
import random

epsilon = 0.2
ads = [0, 0]      # Estimated rewards
clicks = [0, 0]

for _ in range(10):
    if random.random() < epsilon:      # Explore
        ad = random.randint(0, 1)
    else:                              # Exploit
        ad = ads.index(max(ads))

    reward = random.choice([0, 1])     # Simulated click
    clicks[ad] += 1
    ads[ad] += (reward - ads[ad]) / clicks[ad]

print("Estimated Rewards:", ads)
print("Best Ad:", ads.index(max(ads)))
```
