```python
import random

rooms = ["Kitchen", "Hall", "Bedroom", "Charging"]

state = random.choice(rooms)

for i in range(5):
    action = random.choice(["Move", "Charge"])
    if action == "Move":
        state = random.choice(rooms)
    reward = 10 if state == "Charging" else -1
    print(f"Action: {action}, Room: {state}, Reward: {reward}")
```
