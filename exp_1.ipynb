```python
import random

state = "Start"
reward = 0

while state != "End":
    if state == "Start":
        state = "Middle"
    elif state == "Middle":
        state = random.choices(["Win", "Lose"], [0.7, 0.3])[0]
        reward += 10 if state == "Win" else -10
    elif state in ["Win", "Lose"]:
        state = "End"

print("Final State:", state)
print("Reward:", reward)
```
