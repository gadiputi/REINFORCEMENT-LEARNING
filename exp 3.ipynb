```python
import random

state = "Start"

while state != "Goal":
    if state == "Start":
        state = random.choice(["Shelf", "Obstacle"])
    elif state == "Shelf":
        print("Picked Item (+10)")
        state = "Goal"
    else:
        print("Hit Obstacle (-5)")
        state = "Goal"

print("Task Completed")
```
